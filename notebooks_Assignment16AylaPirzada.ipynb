```python
df = df.dropna(subset=["YIELD_KG"])
```

```python
df["AREA_ACRES"] = df["AREA_ACRES"].fillna(df["AREA_ACRES"].mean())
```

```python
df["DISTRICT"] = df["DISTRICT"].fillna("Unknown")
```

```python
df["FERTILIZER_USED"] = df["FERTILIZER_USED"].fillna("No")
```

```python
df["YIELD_VALUE"] = df["YIELD_VALUE"].fillna(df["YIELD_VALUE"].mean())
```

```python
df["HARVEST_DATE"] = pd.to_datetime(df["HARVEST_DATE"], errors="coerce")
```

```python
df.dtypes
```

```python
yield_by_crop = df.groupby("CROP_TYPE").agg({"YIELD_KG": "sum"})
display(yield_by_crop)
```

```python
import matplotlib.pyplot as plt
ax_bar = yield_by_crop["YIELD_KG"].plot(kind="bar",title="Total Yield by Crop Type (KG)",ylabel="Yield (KG)",xlabel="Crop Type",figsize=(8, 5))
```

```python
yield_pivot = df.pivot_table(values="YIELD_KG", index="HARVEST_DATE", aggfunc="sum").sort_index()
```
